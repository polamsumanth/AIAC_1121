**TASK 01**

**INITIAL PROGRAM**

```python
tasks = []

while True:
    print("\n1. Add Task\n2. View Tasks\n3. Exit")
    choice = input("Enter choice: ")
    if choice == '1':
        task = input("Enter task: ")
        tasks.append(task)
    elif choice == '2':
        for t in tasks:
            print(t)
    elif choice == '3':
        break
```

**ENHANCED PROGRAM**

```python
tasks = []
def show_menu():
    print("\n1. Add Task\n2. View Tasks\n3. Delete Task\n4. Exit")
while True:
    show_menu()
    choice = input("Enter choice: ")
    if choice == '1':
        task = input("Enter task: ")
        if task:
            tasks.append(task)
            print("Task added")
        else:
            print("Invalid input")
    elif choice == '2':
        if not tasks:
            print("No tasks available")
        else:
            for i, t in enumerate(tasks, 1):
                print(f"{i}. {t}")
    elif choice == '3':
        try:
            num = int(input("Enter task number: "))
            if 1 <= num <= len(tasks):
                removed = tasks.pop(num - 1)
                print("Removed:", removed)
            else:
                print("Invalid number")
        except:
            print("Enter a valid number")
    elif choice == '4':
        break
    else:
        print("Invalid choice")
```

**STUDENT A INITIAL CODE**

```python
tasks = []

while True:
    print("\n1.Add 2.View 3.Exit")
    ch = input()
    if ch == '1':
        tasks.append(input("Task: "))
    elif ch == '2':
        for t in tasks:
            print(t)
    elif ch == '3':
        break
```

**STUDENT B ENHANCED**

```python
tasks = []

while True:
    print("\n1.Add 2.View 3.Mark Done 4.Exit")
    ch = input()
    if ch == '1':
        task = input("Task: ")
        tasks.append({"task": task, "done": False})
    elif ch == '2':
        for i, t in enumerate(tasks, 1):
            status = "✔" if t["done"] else "✘"
            print(f"{i}. {t['task']} [{status}]")
    elif ch == '3':
        n = int(input("Task number: "))
        if 1 <= n <= len(tasks):
            tasks[n-1]["done"] = True
    elif ch == '4':
        break
```
